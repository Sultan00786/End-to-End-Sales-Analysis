### Data Cleaning and Wrangling

##### 1. Merge with costumers

```python
df_sales.head()
```

```python
df_customers.head()
```

```python
df_sales.merge(
    df_customers, 
    left_on="Customer Name Index", 
    right_on="Customer Index", 
    how="left"
)
```

```python
df_sales= df_sales.merge(
    df_customers, 
    left_on="Customer Name Index", 
    right_on="Customer Index", 
    how="left"
).drop(columns=["Customer Index"])
df_sales.info()
```

##### 2. Merge with products

```python
df_sales.info()
```

```python
df_products.info()
```

```python
df_sales.merge(
    df_products,
    left_on="Product Description Index",
    right_on="Index",
    how="left"
).info()
```

```python
df_sales = df_sales.merge(
    df_products,
    left_on="Product Description Index",
    right_on="Index",
    how="left"
).drop(columns=["Index"])
df_sales.info()
```

##### 3. Merge with region

```python
df_regions.info()
```

```python
df_sales = df_sales.merge(
    df_regions,
    left_on="Delivery Region Index",
    right_on="id",
    how="left"
).drop(columns=["id"])
df_sales.info()
```

```python
df_sales\
    .rename(
        columns={df_sales.columns[14]:"Region Name"}
    )\
    .info()
```

```python
df_sales = df_sales\
    .rename(
        columns={df_sales.columns[14]:"Region Name"}
    )
df_sales.info()
```

##### 4. Merge with state

```python
df_state.info()
```

```python
df_sales.info()
```

```python
df_sales = df_sales.merge(
    df_state,
    left_on="state_code",
    right_on="State Code",
    how="left"
).drop(columns=["State Code", "Region"])
df_sales.info()
```

##### 5. Merge with 2017 buget

```python
df_budgets.info()
```

```python
df_sales.merge(
    df_budgets,
    on='Product Name'
).info()
```

```python
df_sales = df_sales.merge(
    df_budgets,
    on='Product Name'
)
```

#### Remove identical colms

```python
without_dup_col = ~df.columns.duplicated()
df = df.iloc[:,without_dup_col]
df.info()
```

#### Rename the cols

```python
rename_cols = {
    'ordernumber': 'order_number',
    'orderdate': 'order_date',
    'customer names': 'customer_name',
    'product name': 'product_name',
    'order quantity': 'order_quantity',
    'unit price': 'unit_price',
    'line total': 'revenue',
    'total unit cost': 'cost',
    'state_code': 'state_code',
    'state': 'state_name',
    'latitude': 'lat',
    'longitude': 'lon',
    '2017 budgets': 'budget'
}
df = df.rename(columns=rename_cols)
df.info()
```
